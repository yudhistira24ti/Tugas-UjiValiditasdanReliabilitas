```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
df = pd.read_excel(nama_file)

df.columns = [re.sub(r"\s+", " ", str(c)).strip() for c in df.columns]

kolom_butir = [c for c in df.columns if re.match(r"^(.*?) \[", c)]
butir = df[kolom_butir]

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", int(butir.isna().sum().sum()))
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
dimensi = {}
for kolom in butir.columns:
    nama = re.match(r"^(.*?) \[", kolom).group(1).strip(" .")
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")
```

```python
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")

hasil_validitas = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)
    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)
tabel_validitas
```

```python
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
    return ("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
            else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas = []
alpha_dihapus = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)
    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a),
    })
    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(cronbach_alpha(x.drop(columns=kolom)), 3),
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)
display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")
gabung = gabung.merge(tabel_reliabilitas[["Dimensi", "Cronbach's alpha"]], on="Dimensi")
gabung["Tinjau?"] = np.where(gabung["Alpha jika dihapus"] > gabung["Cronbach's alpha"], "Ya", "")

print("Valid (r hitung > r tabel)  :", int((gabung["r hitung"] > r_tabel).sum()), "dari", len(gabung))
print("Valid (corrected r >= 0,30) :", int((gabung["Corrected r"] >= 0.30).sum()), "dari", len(gabung))
gabung[gabung["Tinjau?"] == "Ya"]
```

```python
def uji_buang(nama_dimensi, nomor_butir):
    daftar = dimensi[nama_dimensi]
    x = num[daftar]
    x2 = x.drop(columns=daftar[nomor_butir - 1])
    total = x2.sum(axis=1)
    r_tab = (lambda n: stats.t.ppf(0.975, n-2) / np.sqrt(n-2 + stats.t.ppf(0.975, n-2)**2))(len(x2))
    r_hit = [stats.pearsonr(x2[c], total)[0] for c in x2.columns]
    return {
        "Dimensi": nama_dimensi,
        "Butir dibuang": f"B{nomor_butir}",
        "Alpha sebelum": round(cronbach_alpha(x), 3),
        "Alpha sesudah": round(cronbach_alpha(x2), 3),
        "Butir tersisa": x2.shape[1],
        "Semua butir tersisa valid": all(r > r_tab for r in r_hit),
    }

pd.DataFrame([
    uji_buang("Penggunaan AI pada tugas atau projek kuliah", 4),
    uji_buang("Prokrastinasi akademik", 4),
])
```

```python
sub = num.sample(20, random_state=1)
n2 = len(sub)
t2 = stats.t.ppf(0.975, n2 - 2)
r_tabel2 = t2 / np.sqrt((n2 - 2) + t2**2)
print(f"n = {n2} | r tabel (df={n2-2}) = {r_tabel2:.3f}  (sebelumnya n={n}: {r_tabel:.3f})")

hasil_sub = []
for nama, daftar in dimensi.items():
    x = sub[daftar]
    total = x.sum(axis=1)
    for i, kolom in enumerate(daftar, 1):
        r, _ = stats.pearsonr(x[kolom], total)
        rk, _ = stats.pearsonr(x[kolom], total - x[kolom])
        hasil_sub.append({"Dimensi": nama, "Butir": f"B{i}", "r hitung": round(r, 3),
                          "Corrected r": round(rk, 3), "Valid": r > r_tabel2})
tabel_sub = pd.DataFrame(hasil_sub)
print("Butir valid (n=20):", int(tabel_sub["Valid"].sum()), "dari", len(tabel_sub))
print("Alpha per dimensi (n=20):")
print({nama: round(cronbach_alpha(sub[d]), 3) for nama, d in dimensi.items()})
tabel_sub
```

```python
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")

n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
tabel_keterangan = pd.DataFrame({
    "Parameter": ["Jumlah responden (n)", "Derajat bebas (df)", "r tabel (α=0,05; dua sisi)",
                  "Jumlah butir", "Jumlah dimensi"],
    "Nilai": [n, n - 2, round(r_tabel, 3), num.shape[1], len(dimensi)],
})

daftar_butir = []
for nama, daftar in dimensi.items():
    for i, kolom in enumerate(daftar, 1):
        daftar_butir.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "Pernyataan": kolom.split("[", 1)[1].rstrip("]"),
        })
tabel_daftar_butir = pd.DataFrame(daftar_butir)

nama_output = "hasil_uji_validitas_reliabilitas.xlsx"
with pd.ExcelWriter(nama_output, engine="openpyxl") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)
    tabel_keterangan.to_excel(w, sheet_name="Keterangan", index=False)
    tabel_daftar_butir.to_excel(w, sheet_name="Daftar Butir", index=False)

    for ws in w.book.worksheets:
        for kol in ws.columns:
            lebar = max(len(str(c.value)) if c.value is not None else 0 for c in kol)
            ws.column_dimensions[kol[0].column_letter].width = min(max(12, lebar + 2), 60)

files.download(nama_output)
```
